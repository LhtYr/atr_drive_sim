```python
# 加载数据
from plot import setup_style, load_data, plot_series, render_plots, get_column

setup_style()
data = load_data('../results/simulation.csv')
t = data['t']
print(f'{len(data)} rows, columns: {list(data.dtype.names)}')
```

```python
# 批量绘制全部图表并保存到 results/figures/（与 make plot 等效）
render_plots(data, outdir='../results/figures', show=True)
```

```python
# 示例：放大观察任意时间窗（改 t0/t1 即可）
t0, t1 = 0.0, 0.05
mask = (t >= t0) & (t <= t1)
plot_series(t[mask], [data['id'][mask], data['iq'][mask]], ['id (A)', 'iq (A)'],
            f'dq 电流响应（{t0}–{t1} s）', ylabel='Current (A)', show=True)
```
